```python
import torch
```

```python
x=torch.tensor(4.0,requires_grad=True)
```

```python
y=x**2
```

```python
z=torch.sin(y)
```

```python
x
```

```python
y
```

```python
y.backward(retain_graph=True)
```

```python
#Backward propgation value
x.grad
```

```python
z
```

```python
#it will cancel the previous gradient decent value to compute the future added functions in the graph which means clearing gradient
x.grad.zero_()
```

```python
z.backward()
```

```python
x.grad
```
